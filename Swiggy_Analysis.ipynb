```python
rating_available = (df["Rating"] > 0).sum()
```

```python
rating_unavailable = (df["Rating"] == 0).sum()
```

```python
rating_available_pct = rating_available / len(df) * 100

print("Rating Available:", rating_available)
print("Rating Unavailable:", rating_unavailable)
print("Rating Availability:", round(rating_available_pct, 2), "%")
```

```python
rating_count_available = (df["Rating Count"] > 0).sum()
rating_count_unavailable = (df["Rating Count"] == 0).sum()
rating_count_pct = rating_count_available / len(df) * 100

print("Rating Count Available:", rating_count_available)
print("Rating Count Unavailable:", rating_count_unavailable)
print("Rating Count Availability:", round(rating_count_pct, 2), "%")
```

```python
state_analysis = (
    df.groupby("State")
      .size()
      .sort_values(ascending=False)
)

print(state_analysis)
```

```python
top_states = (
    df["State"]
    .value_counts()
    .head(10)
)

print(top_states)
```

```python
top_cities = (
    df["City"]
    .value_counts()
    .head(10)
)

print(top_cities)
```

```python
top_restaurants = (
    df["Restaurant Name"]
    .value_counts()
    .head(10)
)

print(top_restaurants)
```

```python
top_categories = (
    df["Category"]
    .value_counts()
    .head(10)
)

print(top_categories)
```

```python
avg_price_state = (
    df.groupby("State")["Price (INR)"]
      .mean()
      .sort_values(ascending=False)
)

print(avg_price_state)
```

```python
avg_rating_state = (
    df.groupby("State")["Rating"]
      .mean()
      .sort_values(ascending=False)
)

print(avg_rating_state)
```

```python
df["Price Bucket"] = pd.cut(
    df["Price (INR)"],
    bins=[0, 250, 500, 1000, float("inf")],
    labels=["Below ₹250", "₹250-₹499", "₹500-₹1,000", "Above ₹1,000"],
    right=False
)

print(df["Price Bucket"].value_counts().sort_index())
```

```python
print("\n--- Average Rating by State (Rated Records Only) ---")

rated_df = df[df["Rating"] > 0]

avg_rating_state_rated = (
    rated_df.groupby("State")["Rating"]
    .mean()
    .sort_values(ascending=False)
    .round(2)
)

print(avg_rating_state_rated)
```

```python
print("\n--- Price Bucket Analysis ---")

df["Price Bucket"] = pd.cut(
    df["Price (INR)"],
    bins=[0, 250, 500, 1000, float("inf")],
    labels=[
        "Below ₹250",
        "₹250-₹499",
        "₹500-₹1,000",
        "Above ₹1,000"
    ],
    right=False
)

price_bucket_analysis = (
    df["Price Bucket"]
    .value_counts()
    .sort_index()
)

print(price_bucket_analysis)
```

```python
print("\n--- Average Rating by Price Bucket ---")

avg_rating_price_bucket = (
    df.groupby("Price Bucket", observed=False)["Rating"]
    .mean()
    .round(2)
)

print(avg_rating_price_bucket)
```

```python
print("\n--- Top 10 Restaurants by Average Price ---")

restaurant_avg_price = (
    df.groupby("Restaurant Name")["Price (INR)"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .round(2)
)

print(restaurant_avg_price)
```

```python
print("\n--- Top 10 Restaurants by Average Rating ---")

restaurant_avg_rating = (
    rated_df.groupby("Restaurant Name")["Rating"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .round(2)
)

print(restaurant_avg_rating)
```

```python
print("\n--- Top 10 Cities by Average Price ---")

city_avg_price = (
    df.groupby("City")["Price (INR)"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .round(2)
)

print(city_avg_price)
```

```python
with pd.ExcelWriter("Swiggy_Python_Analysis.xlsx") as writer:

    top_states.to_excel(writer, sheet_name="Top States")
    top_cities.to_excel(writer, sheet_name="Top Cities")
    top_restaurants.to_excel(writer, sheet_name="Top Restaurants")
    top_categories.to_excel(writer, sheet_name="Top Categories")
    avg_price_state.to_excel(writer, sheet_name="Avg Price State")
    avg_rating_state_rated.to_excel(writer, sheet_name="Avg Rating State")
    price_bucket_analysis.to_excel(writer, sheet_name="Price Buckets")
    avg_rating_price_bucket.to_excel(
        writer,
        sheet_name="Rating Price Bucket"
    )

print("\nAnalysis completed successfully!")
print("Output file: Swiggy_Python_Analysis.xlsx")
```
